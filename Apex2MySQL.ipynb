```python
import mysql.connector
import pandas as pd
from getpass import getpass


conn = mysql.connector.connect(
    host="localhost",
    user="root",
    password=getpass("ansh123"),
    database="intership"
)

print("Connected Successfully!")
```

```python
# 1 Sales by Gende( Male VS Female )
Query1="""select gender ,Round(sum(Total_sales),0) as Total_sales from  apex_planet group by gender;"""
df=pd.read_sql(Query1, conn)
df.index = range(1, len(df) + 1)
df
```

```python
#-- 2. Top-Selling Category by City
Query2="""with T as (select city ,category,total_sales,dense_rank() over (partition by city order by total_sales desc) as rnk from apex_planet)
select city,category,total_sales from T where rnk=1 ;"""
df=pd.read_sql(Query2, conn)
df.index = range(1, len(df) + 1)
df
```

```python
#-- 3. Top 3 Products by Total Sales
Query3="""select product,count(*) as Total_customers ,round(sum(Total_sales)) as Total_sales 
from apex_planet group by product order by round(sum(Total_sales)) desc limit 3;"""
df=pd.read_sql(Query3, conn)
df.index = range(1, len(df) + 1)
df
```

```python
#-- 4. Top 10 Customers by Total Purchases
Query4="""select customer_name,round(sum(Total_sales)) from apex_planet group by customer_name order by  Round(sum(Total_sales),0) desc limit 10;"""
df=pd.read_sql(Query4, conn)
df.index = range(1, len(df) + 1)
df
```

```python
#-- 5. Top 3 Categories by Gender
Query5="""with a as (select gender,category,round(sum(total_sales),0) as total_sales from apex_planet group by gender,category),
b as (select gender,category,Total_sales,dense_rank() over (partition by gender order by total_sales desc ) as Rnk
 from a order by gender,Rnk) 
 select * from b where rnk<=3;"""
df=pd.read_sql(Query4, conn)
df.index = range(1, len(df) + 1)
df
```
